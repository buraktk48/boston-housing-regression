```python
y_train_log = np.log1p(y_train)
```

```python
from sklearn.linear_model import Ridge
```

```python
X_train_ridge = X_train.copy()
X_test_ridge = X_test.copy()

X_train_ridge["INDUS_NOX"] = X_train_ridge["INDUS"] * X_train_ridge["NOX"]
X_train_ridge["AGE_LSTAT"] = X_train_ridge["AGE"] * X_train_ridge["LSTAT"]

X_test_ridge["INDUS_NOX"] = X_test_ridge["INDUS"] * X_test_ridge["NOX"]
X_test_ridge["AGE_LSTAT"] = X_test_ridge["AGE"] * X_test_ridge["LSTAT"]
```

```python
X_train_ridge.columns
```

```python
numerical_cols = ['CRIM', 'ZN', 'INDUS', 'NOX', 'RM', 'AGE', 'DIS', 'TAX', 'PTRATIO', 'B', 'LSTAT',"AGE_LSTAT","INDUS_NOX"]
scaler = StandardScaler()

X_train_ridge[numerical_cols] = scaler.fit_transform(X_train_ridge[numerical_cols])
X_test_ridge[numerical_cols] = scaler.transform(X_test_ridge[numerical_cols])

ridge = Ridge(alpha=1.0)
ridge.fit(X_train_ridge, y_train_log)

y_pred_ridge = ridge.predict(X_test_ridge)

y_pred_orig = np.expm1(y_pred_ridge)

r2_log = r2_score(y_test, y_pred_orig)
rmse_log = mean_squared_error(y_test, y_pred_orig) ** 0.5

print(f"--- LOG DÖNÜŞÜMLÜ MODEL SONUÇLARI ---")
print(f"Yeni R² Skoru: {r2_log:.4f}")
print(f"Yeni RMSE: {rmse_log:.4f}")
```

**Çoklu doğrusallık sorununu önlemek için Ridge modelini kullanacağız**

```python
df.columns
```

**Since our target variable does not follow a normal distribution, we take its logarithm**

```python
from sklearn.ensemble import RandomForestRegressor


rf_model = RandomForestRegressor(random_state=42)


rf_model.fit(X_train_ridge, y_train_log)


y_pred_rf_log = rf_model.predict(X_test_ridge)


y_pred_rf_orig = np.expm1(y_pred_rf_log)


r2_rf = r2_score(y_test, y_pred_rf_orig)
rmse_rf = mean_squared_error(y_test, y_pred_rf_orig) ** 0.5

print(f"--- RANDOM FOREST MODEL SONUÇLARI ---")
print(f"Yeni R² Skoru: {r2_rf:.4f}")
print(f"Yeni RMSE: {rmse_rf:.4f}")
```

```python

y_train_pred_log = rf_model.predict(X_train_ridge)
y_train_pred_orig = np.expm1(y_train_pred_log)
r2_train = r2_score(y_train,y_train_pred_orig)

print(f"Eğitim Seti R² Skoru: {r2_train:.4f}")
print(f"Test Seti R² Skoru: {r2_rf:.4f}")
```

**It looks like there's a bit of overfitting, so we're fine-tuning the model.**

```python
from sklearn.model_selection import GridSearchCV
```

```python
rf_params = {'max_depth': [5,10,15,20,None],
            'max_features': [0.6,0.8,1.0],
            "min_samples_split": [2, 5, 10],
             "min_samples_leaf": [1, 2, 4],
            'n_estimators' : [50,100, 200, 500]}


rf_model = RandomForestRegressor(random_state = 42)

rf_cv_model = GridSearchCV(rf_model, 
                           rf_params, 
                           cv = 5, 
                            n_jobs = -1)

rf_cv_model.fit(X_train_ridge, y_train)
```

```python
y_pred_log_rf_cv_model = rf_cv_model.predict(X_test_ridge)
```

```python
y_pred_rf_cv_model = np.expm1(y_pred_log_rf_cv_model)
```

```python
np.sqrt(mean_squared_error(y_test, y_pred_rf_cv_model))
```

```python
parametreler =rf_cv_model.best_params_
```

```python
parametreler
```

```python
rf_tuned = RandomForestRegressor(max_depth=15,max_features=0.6,min_samples_leaf=1,min_samples_split=2,n_estimators=100)
```

```python
rf_tuned.fit(X_train_ridge,y_train)
```

```python
y_pred_rf_tuned = rf_tuned.predict(X_test_ridge)
```

```python
np.sqrt(mean_squared_error(y_test,y_pred_rf_tuned))
```

```python
r2_rf = r2_score(y_test,y_pred_rf_tuned)
r2_rf
```

```python
y_pred_rf_tuned_train = rf_tuned.predict(X_train_ridge)
r2_egitim = r2_score(y_train,y_pred_rf_tuned_train)
r2_egitim
```
